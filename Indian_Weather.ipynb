# Indian Weather Forecasting: 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
weather_astro=pd.read_csv('astro_data.csv')
weather_forecast=pd.read_csv('forecast_data.csv')
weather_location=pd.read_csv('location_data.csv')
weather_location = weather_location.rename(columns={'name': 'city'})
astro_forecast = pd.merge(weather_astro,weather_forecast,on=['city'],how='left')
astro_forecast.head(2)
```

```python
weather = pd.merge(astro_forecast,weather_location,on=['city'],how='left')
weather.head(2)
```

## Graph Visualization : 

## 1. Distribution and Variation Analysis :

```python
# Q.1 What is the overall distribution of temperature readings (temp_c) across the entire dataset? Histplot

temp_readings=weather['temp_c'].describe()
temp_readings
```

```python
plt.figure(figsize=(12,6))
sns.histplot(data=weather, x='temp_c',kde=True,bins=30,color='skyblue')
plt.title('Distribution of Temprature Readings')
plt.xlabel('temp_c')
```

#### Conclusion :
* The temperature readings in the dataset are around an average of about $\mathbf{25^{\circ}\text{C}}$.
* Most of the recorded temperatures fall between $\mathbf{22.5^{\circ}\text{C}}$ and $\mathbf{28.1^{\circ}\text{C}}$.
* While most readings are within this comfortable, warm range, the dataset also includes some much colder temperatures, with the minimum reading being $\mathbf{1.5^{\circ}\text{C}}$.

```python
# Q.2 How does the wind speed (wind_kph) vary over the entire time period recorded? Box Plot or Violin Plot

weather['time'] = pd.to_datetime(weather['time'])
weather['hour'] = weather['time'].dt.hour
wind_speed = weather.groupby('hour')['wind_kph'].agg(['mean', 'median', 'std', 'min', 'max'])
print("Summary of Wind Speed (wind_kph) by Hour:")
wind_speed
```

```python
weather['time'] = pd.to_datetime(weather['time'])
weather['hour'] = weather['time'].dt.hour
plt.figure(figsize=(14,6))
sns.violinplot(data=weather,x='hour',y='wind_kph',hue='hour',palette='RdBu')
plt.title('Wind speed over Time Period')
plt.xlabel('Hour')
plt.ylabel('Wind_Speed_kph')
plt.legend()
plt.tight_layout()
```

#### Conclusion: 
* Wind speeds are generally highest during the day, peaking around ($\mathbf{12:00 \text{ PM}}$) with an average of about $\mathbf{10 \text{ kph}}$.
* Wind speeds are typically lowest in the early morning (around $\mathbf{6:00 \text{ AM}}$), averaging close to $\mathbf{7 \text{ kph}}$.
* The spread and maximum values of wind speed are also highest during the middle of the day, suggesting more turbulent conditions when winds are at their strongest.

```python
# Q.3 What is the frequency of each unique moon_phase?  Bar Chart

moon_pcounts = weather['moon_phase'].value_counts().reset_index()
moon_pcounts.columns = ['moon_phase', 'frequency']
moon_pcounts
```

```python
plt.figure(figsize=(12,6))
m=sns.barplot(data=moon_pcounts,x='moon_phase',y='frequency',hue='moon_phase',palette='rocket')
for i in m.containers:
    m.bar_label(i)
plt.title('Frequency of Each Moon Phase',fontsize=20)
plt.xticks(rotation=35,fontsize=10)
plt.xlabel('Moon_Phase',fontsize=15)
plt.ylabel('Frequency',fontsize=15)
plt.tight_layout()
```

#### Conclusion :
* The distribution of moon phases in the dataset is highly uneven, with Waxing Gibbous being the most frequent phase by a large margin (924 readings).
* The First Quarter and Full Moon phases have the exact same, lowest frequency ($\mathbf{154}$ readings each).
* This suggests that the data collected multiple days where the moon was most of the time in the Waxing Gibbous phase, which covers a significant portion of the lunar cycle.

## 2. Time-Series and Trend Analysis :


```python
# Q.1 How does the average temperature (temp_c) change hourly throughout the day?  lineplot

weather['time'] = pd.to_datetime(weather['time'])
weather['hour'] = weather['time'].dt.hour
avg_temp_ph = weather.groupby('hour')['temp_c'].mean().reset_index()
avg_temp_ph.columns = ['hour', 'avg_temp_c']
avg_temp_ph
```

```python
plt.figure(figsize=(12,6))
sns.lineplot(data=avg_temp_ph,x='hour',y='avg_temp_c',linestyle='solid',linewidth=2,marker='*',markersize=20,color='skyblue')
plt.title('Average Temp (temp_c) Hourly Throughout The Day',fontsize=20)
plt.xlabel('Hour',fontsize=15)
plt.ylabel('Average Temprature',fontsize=15)
plt.grid()
plt.tight_layout()
```

#### Conclusion :
* The temperature is lowest in the early morning, averaging around $\mathbf{22.2^{\circ} \text{C}}$ at 6:00 AM.
* The temperature gradually drops overnight, reaching its minimum average of $\mathbf{22.20^{\circ} \text{C}}$ around 6:00 AM, just before sunrise.
* The temperature rises much faster during the day (from 6:00 AM to 12:00 PM) than it falls during the night (from 06:00 PM to 06:00 AM).

```python
# Q.2 Is there a relationship between humidity and dewpoint_c over time?  Line Plot (Dual Axis or Two lines on one plot)

weather['time_only'] = pd.to_datetime(weather['time']).dt.strftime('%H')
```

```python
plt.figure(figsize=(10,5))

sns.lineplot(data=weather, x='time_only', y='humidity', label='Humidity')
sns.lineplot(data=weather, x='time_only', y='dewpoint_c', label='Dew Point (°C)')

plt.title("Humidity vs Dew Point Over Time")
plt.xlabel("Time")
plt.ylabel("Values")
plt.xticks(rotation=45)
plt.show()
```

```python
fig, ax1 = plt.subplots(figsize=(10,5))

sns.lineplot(data=weather, x='time_only', y='humidity', ax=ax1, color='blue', label='Humidity')
ax1.set_xlabel("Time")
ax1.set_ylabel("Humidity", color='blue')
ax1.tick_params(axis='y', labelcolor='blue')
ax1.tick_params(axis='x', rotation=45)

ax2 = ax1.twinx()
sns.lineplot(data=weather, x='time_only', y='dewpoint_c', ax=ax2, color='red', label='Dew Point (°C)')
ax2.set_ylabel("Dew Point (°C)", color='red')
ax2.tick_params(axis='y', labelcolor='red')

plt.title("Humidity vs Dew Point Over Time (Dual Axis)")
plt.show()
```

#### Conclusion :
* From the graph, we can see that humidity and dew point move in a similar pattern over time.
* When the dew point increases, the humidity also increases, and when the dew point decreases, the humidity drops.
* This means there is a positive relationship between humidity and dew point.

```python
# Q.3 How do gust_kph and wind_kph compare hourly? lineplot
# Gust_kph = sudden, short bursts of stronger wind

plt.figure(figsize=(10,5))
sns.lineplot(data=weather, x='hour', y='wind_kph',label='Wind Speed kph')
sns.lineplot(data=weather, x='hour', y='gust_kph',label='Gust Speed kph')
plt.title('Comparison between Gust and Wind Speed Hourly')
plt.xlabel('Hours')
plt.ylabel('Speed_kph')
plt.tight_layout()
```

#### Conclusion :
* Gust speed (gust_kph) is generally higher than normal wind speed (wind_kph) for each hour.
* Both follow a similar pattern over time, but gusts show stronger spikes.
* Wind_kph = normal, continuous wind speed.
* Gust_kph = sudden, short bursts of stronger wind.
* If normal wind is 20 kph, a gust may suddenly jump to 35 kph for a few seconds.

## 4. Relationship and Correlation Analysis :

```python
weather.columns
```

```python
# Q.1 What is the correlation between feelslike_c and actual temp_c? scatterplot

weather['temp_c'].corr(weather['feelslike_c'])
```

```python
plt.figure(figsize=(12,6))
sns.scatterplot(data=weather,x='temp_c',y='feelslike_c',hue='temp_c')
plt.title('Correlation Between feelslike_c and temp_c')
plt.xlabel('temp_c')
plt.ylabel('feelslike_c')
plt.tight_layout()
```

```python
# Q.2 Is there a relationship between moon_illumination and humidity? scatterplot

weather['humidity'].corr(weather['moon_illumination'])
plt.figure(figsize=(12,6))
sns.scatterplot(data=weather,x='humidity',y='moon_illumination')
plt.title('Relationship Between Moon_illumination and Humidity')
plt.xlabel('Humidity')
plt.ylabel('Moon_illumination')
plt.tight_layout()
```

#### Conclusion :
* There is no strong relationship between moon_illumination and humidity.
* The points are widely scattered, and the correlation value is close to zero, meaning moon illumination does not affect humidity.

```python
# Q.3 How do all numerical variables correlate with each other? Heatmap

num_var = weather.corr(numeric_only=True)

plt.figure(figsize=(12,6))
sns.heatmap(num_var,cmap='coolwarm',linewidths=0.5)

plt.title("Correlation Heatmap of Numerical Variables")
plt.tight_layout()
plt.show()
```

#### Conclusion :
* Positive values (closer to +1) show strong positive correlation,
* negative values (closer to −1) show strong negative correlation, and 
* values near 0 indicate weak or no relationship.
* Darker or stronger colors show higher correlation, while lighter colors indicate weaker relationships.
* From the heatmap, we can observe that:
  * Some weather variables show clear patterns, such as wind, gust, temperature, and humidity having noticeable color gradients.
  * Other variables show very weak or almost no correlation, as indicated by very light shades.
  * Overall, the heatmap provides a quick visual understanding of which numerical features move together and which behave independently.

### Q.# What are the average sunrise and sunset times for each state, and which states experience earlier sunrise and later sunset?


```python
weather['sunrise'] = pd.to_datetime(weather['sunrise'])
state_sunrise = weather.groupby('state_x')[['sunrise']].mean().reset_index()
state_sunrise
```

```python
# Visualize Sunrise & Sunset by State
# 1.Sunrise time bar chart

plt.figure(figsize=(12,6))
s=sns.barplot(data=state_sunrise,y='state_x',x='sunrise',hue='state_x')
for i in s.containers:
    s.bar_label(i,fontsize=10)
plt.title("Average Sunrise Time by State")
plt.xlabel("State")
plt.ylabel("Sunrise Time")
plt.xticks(rotation=90)
```

```python
weather['sunset'] = pd.to_datetime(weather['sunset'])
state_sunset = weather.groupby('state_x')[['sunset']].mean().reset_index()
```

```python
# 2.Sunset time bar chart

plt.figure(figsize=(12,6))
s=sns.barplot(data=state_sunset,y='state_x',x='sunset',hue='state_x')
for i in s.containers:
    s.bar_label(i,fontsize=10)
plt.title("Average Sunset Time by State")
plt.xlabel("State")
plt.ylabel("Sunset Time")
plt.xticks(rotation=90)
```

#### Conclusion :
* The analysis shows that sunrise and sunset times vary across states.
* Some states experience earlier sunrise, meaning daylight starts sooner, while others have later sunrise.
* Similarly, states with a later sunset enjoy longer daylight hours compared to those where the sun sets earlier.
* Overall, the pattern highlights how geographical location influences the length of the day in each state.
